```python
from re import compile
tok = np.array([b'\n !"%&\'()*+,-./0123456789:<=>@[]^abcdefghijklmnopqrstuvwxyz{|}~']).view('B')
utok = np.full(2<<32-1,255,'B')
utok[tok] = range(len(tok))
utok[ord('\t')] = utok[ord(' ')]
utok[ord('A'):ord('Z')+1] = utok[ord('a'):ord('z')+1]
nn = compile('\n{2,}')
cm = compile('\\s*(#.*?)?\n')
a = open('py.txt', errors='ignore')
o = open(f'py.{len(tok)}','wb')
while t := a.read(1048576):
    t = utok[np.array([nn.sub('\n',cm.sub('\n',t)).replace('    ',' ')]).view('I')]
    t[t!=255].tofile(o)
tok.tofile(f'py.{len(tok)}tok')
```
