```python
__import__('importlib').reload(__import__('nn'))
from nn import *
m = 256
def Cnv(i=m, h=m, r=7):
    return Sequential(
        Linear(i, h, 0),
        T(Conv1d(h, h, r, padding=r//2, groups=h, bias=0)),
        LayerNorm(h, elementwise_affine=0),
    )
def At(i=m, h=m//16, H=m//16, g=16, G=1, r=8):
    return Sequential(
        Res(
            Attn(i, h, H, g, G, r),
            Linear(G*g*(H or i), i, 0),
        ),
        LayerNorm(i, elementwise_affine=0),
    )
def Ff(i=m, h=m):
    return Sequential(
        Res(
            SwiGLU(i, h),
            Linear(h, i, 0),
        ),
        LayerNorm(i, elementwise_affine=0),
    )
def U(*_, i=m, h=m, s=2):
    return Res(
        Linear(i, h, 0),
        T(MaxPool1d(s)),
        *_,
        T(ConvTranspose1d(h, h, s, s, groups=h, bias=0)),
        Linear(h, i, 0),
    )
m = torch.compile(Sequential(
    Embedding(65536, m),
    Cnv(),
    Ff(),
    U(
        Cnv(),
        Ff(),
        U(
            At(),
            Ff(),
            U(
                Cnv(),
                Ff(),
                U(
                    Cnv(),
                    Ff(),
                    U(
                        At(),
                        Ff(),
                    ),
                    Cnv(),
                    Ff(),
                ),
                Cnv(),
                Ff(),
            ),
            At(),
            Ff(),
        ),
        Cnv(),
        Ff(),
    ),
    Cnv(),
    Ff(),
    Linear(m, 65536),
), fullgraph=True, dynamic=False).to('cuda',torch.bfloat16)
o = torch.optim.Adam(m.parameters(), 1/512)
a = np.lib.stride_tricks.sliding_window_view(np.memmap('py.65536', 'H', 'r'), 4096)
plot = Plot(2, 10, 10)
for y in range(1000000):
    t = torch.tensor(a[np.random.randint(0,len(a),2)], dtype=torch.long, device='cuda')
    x = t.put(torch.randint(t.numel(),[t.numel()//2],device='cuda'),torch.randint(65536,[t.numel()//2],device='cuda'))
    y = m(x)
    l = functional.cross_entropy(y.mT, t)
    (l*1024).backward()
    o.step()
    o.zero_grad()
    plot(l, lambda:(y.argmax(-1)==t).sum()/t.numel())
```
