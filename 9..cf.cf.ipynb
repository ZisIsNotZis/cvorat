```python
from nn import *
class DyT(Module):
    def __init__(s):
        super().__init__()
        s._ = Parameter(torch.ones(()))
    def forward(s, _):
        return (_*s._).tanh()
def Cv(i, h=0, r=(7,0)):
    return Res(
        Linear(i, h or i, 0),
        T(Conv1d(h or i, h or i, sum(r)+1, padding=max(r), groups=h or i, bias=0)),
        F(lambda _:_[...,max(r[1]-r[0],0):min(r[1]-r[0],0)or None,:]if r[1]!=r[0]else _),
        Linear(h or i, i, 0),
    )
def At(i, h=0, H=0, g=1, G=1, r=(7,0)):
    return Res(
        RMSNorm(i, elementwise_affine=0),
        Attn(i, h, H, g, G, r),
        Linear(G*g*(H or i), i, 0),
    )
def Ff(i, h=0):
    return Res(
        RMSNorm(i, elementwise_affine=0),
        SwiGLU(i, h or i),
        Linear(h or i, i, 0),
    )
def U(i, h=0, s=2, S=-1):
    return lambda*_:Res(
        Linear(i, h or i, 0),
        T(MaxPool1d(s)),
        *_,
        T(ConvTranspose1d(h or i, h or i, s, s, output_padding=max(S,0), groups=h or i, bias=0)),
        F(lambda _:_[...,S:,:]if S>0 else functional.pad(_[...,:S,:],(0,0,-S,0))if S else _),
        Linear(h or i, i, 0),
    )
```

```python
m = 256
m = Sequential(
    Embedding(65536, m),
    Add(
        Seq(
            Cv(m),
            Ff(m),
            U(m)(
                Cv(m),
                Ff(m),
            ),
            Cv(m),
            Ff(m),
            F(lambda _:functional.pad(_[...,:-1,:],(0,0,1,0))),
            Linear(m, m),
        ),
        Seq(
            Cv(m,r=(0,7)),
            Ff(m),
            U(m,S=1)(
                Cv(m,r=(0,7)),
                Ff(m),
            ),
            Cv(m,r=(0,7)),
            Ff(m),
            F(lambda _:functional.pad(_[...,1:,:],(0,0,0,1))),
            Linear(m, m, 0),
        ),
    ),
    SiLU(),
    Linear(m, 65536),
).to('cuda',torch.bfloat16)
print(m)
m = torch.compile(m, fullgraph=True, dynamic=False)
o = torch.optim.Adam(m.parameters(), 1/1024)
a = torch.from_numpy(np.memmap('../../py65536.bin', mode='r'))
a = a.as_strided_((len(a)//512-1,1024),(512,1))
plot = Plot(2)
for i in range(1, 100):
    for _ in range(10000):
        t = a[torch.randint(len(a),[4])].to('cuda',torch.long)
        x = t.put(torch.randint(t.numel(),[t.numel()//2],device='cuda'),torch.randint(65536,[t.numel()//2],device='cuda'))
        y = m(x)
        l = functional.cross_entropy(y.mT, t)
        (l*1024).backward()
        o.step()
        o.zero_grad()
        plot(l, lambda:(y.argmax(-1)==t).sum()/t.numel())
```
